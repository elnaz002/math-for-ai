## Sample Space and Events

- **Sample space** $\Omega$: the set of all possible outcomes
- **Event** $A$: a subset of $\Omega$
- **Probability** $P(A)$: a number in $[0, 1]$ satisfying:
  - $P(\Omega) = 1$
  - $P(\emptyset) = 0$
  - For disjoint events: $P(A \cup B) = P(A) + P(B)$

**Example:** rolling a die
- $\Omega = \{1, 2, 3, 4, 5, 6\}$
- $A = \{$even numbers$\} = \{2, 4, 6\}$
- $P(A) = 3/6 = 0.5$

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
```

```python
np.random.seed(42)

rolls = np.random.randint(1, 7, size=10_000)

even_count = np.sum(rolls % 2 == 0)
print(f"P(even) empirical: {even_count / len(rolls):.4f}")
print(f"P(even) theoretical: 0.5")

counts = Counter(rolls)
for face in range(1,7):
  print(f"P{face} = {counts[face]/len(rolls):.4f}")
```

## Conditional Probability

The probability of $A$ **given that** $B$ has occurred:

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}, \quad P(B) > 0$$

**Intuition:** restricting the sample space to $B$.

**Multiplication rule:**
$$P(A \cap B) = P(A \mid B) \cdot P(B) = P(B \mid A) \cdot P(A)$$

```python
rolls = np.random.randint(1, 7, size=100_000)

B = rolls % 2 == 0
A = rolls == 6

# P(A | B) = P(A and B) / P(B)
p_A_and_B = np.mean(A & B)
p_B = np.mean(B)
p_A_given_B = p_A_and_B / p_B

print(f"P(A | B) empirical = {p_A_given_B:.4f}")
print(f"P(A | B) theoretical = 1/3 ≈ {1/3:.4f}")
```

## Bayes' Theorem

The most important formula in probabilistic ML:

$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$

**Terms:**
- $P(A)$ — **prior**: what we believed before seeing data
- $P(B \mid A)$ — **likelihood**: how likely the data is under $A$
- $P(A \mid B)$ — **posterior**: updated belief after seeing data
- $P(B)$ — **evidence**: normalizing constant

**Why it matters for ML:**
- Naive Bayes classifier
- Bayesian inference
- Generative models (VAE, diffusion)
- Any time we update beliefs with data


```python
# Classic example: disease test
# Disease prevalence: 1% of population
# Test sensitivity: P(positive | disease) = 99%
# Test specificity: P(negative | healthy) = 95%
#
# Question: if you test positive, what's P(disease | positive)?

p_disease = 0.01
p_healthy = 0.99

p_pos_given_disease = 0.99   
p_pos_given_healthy = 0.05   


p_pos = p_pos_given_disease * p_disease + p_pos_given_healthy * p_healthy
p_disease_given_pos = (p_pos_given_disease * p_disease) / p_pos

print(f"P(positive) = {p_pos:.4f}")
print(f"P(disease | positive) = {p_disease_given_pos:.4f}")
print(f"→ Only {p_disease_given_pos*100:.1f}% chance you actually have it!")
```

```python
np.random.seed(0)
N = 1_000_000


has_disease = np.random.rand(N) < p_disease


test_positive = np.where(
    has_disease,
    np.random.rand(N) < p_pos_given_disease,   
    np.random.rand(N) < p_pos_given_healthy    
)

pos_mask = test_positive
p_empirical = np.mean(has_disease[pos_mask])
print(f"P(disease | positive) simulated = {p_empirical:.4f}")
print(f"P(disease | positive) Bayesian  = {p_disease_given_pos:.4f}")
```
